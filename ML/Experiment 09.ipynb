# Implementation of K-Means Clustering

**Machine Learning Lab**

This notebook contains the objective, implementation, evaluation, observations, and result.

## Aim
Apply K-Means, an unsupervised learning algorithm, to group observations into clusters. The Wine dataset is used; class labels are not used during clustering.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

data=load_wine(as_frame=True)
X=data.data
X_scaled=StandardScaler().fit_transform(X)
inertias=[]
silhouettes=[]
ks=range(2,8)
for k in ks:
    km=KMeans(n_clusters=k,random_state=42,n_init=10)
    labels=km.fit_predict(X_scaled)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(X_scaled,labels))
display(pd.DataFrame({"k":list(ks),"Inertia":inertias,"Silhouette":silhouettes}))
```

```python
plt.figure(figsize=(7,4))
plt.plot(list(ks),inertias,marker="o")
plt.xlabel("Number of clusters (k)"); plt.ylabel("Inertia")
plt.title("Elbow Method"); plt.show()
plt.figure(figsize=(7,4))
plt.plot(list(ks),silhouettes,marker="o")
plt.xlabel("Number of clusters (k)"); plt.ylabel("Silhouette score")
plt.title("Silhouette Score by k"); plt.show()
```

```python
kmeans=KMeans(n_clusters=3,random_state=42,n_init=10)
clusters=kmeans.fit_predict(X_scaled)
X_2d=PCA(n_components=2).fit_transform(X_scaled)
plt.figure(figsize=(8,5))
plt.scatter(X_2d[:,0],X_2d[:,1],c=clusters,cmap="viridis",s=35)
plt.xlabel("Principal Component 1"); plt.ylabel("Principal Component 2")
plt.title("K-Means Clusters visualized with PCA"); plt.show()
print("Silhouette score for k=3:",silhouette_score(X_scaled,clusters))
```

## Observations and Result
K-Means assigns each observation to the nearest cluster centroid and iteratively updates centroids. Inertia measures within-cluster squared distances; the elbow method is a heuristic for choosing k. The silhouette score measures cluster separation. PCA is used only to visualize the high-dimensional clusters in two dimensions. K-Means clustering was implemented.
