# Dimensionality Reduction using PCA

**Machine Learning Lab**

This notebook contains the objective, implementation, evaluation, observations, and result.

## Aim
Apply Principal Component Analysis (PCA) to reduce the number of features while retaining as much variance as possible. The Wine dataset is used.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

data = load_wine(as_frame=True)
X = data.data
y = data.target
print("Original shape:",X.shape)
X_scaled = StandardScaler().fit_transform(X)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
print("Reduced shape:",X_pca.shape)
print("Explained variance ratio:",pca.explained_variance_ratio_)
print("Total variance retained:",pca.explained_variance_ratio_.sum())
```

```python
pca_df = pd.DataFrame(X_pca,columns=["PC1","PC2"])
pca_df["Class"] = y.map(dict(enumerate(data.target_names)))
display(pca_df.head())
for label, group in pca_df.groupby("Class"):
    plt.scatter(group.PC1,group.PC2,label=label,alpha=0.75)
plt.xlabel("Principal Component 1"); plt.ylabel("Principal Component 2")
plt.title("Wine Dataset after PCA"); plt.legend(); plt.show()
```

## Observations and Result
PCA creates new orthogonal features called principal components. The first two components retain the displayed fraction of total variance, which may be less than 100%. Scaling is important because PCA is sensitive to feature magnitude. Dimensionality was reduced from the original feature count to two components.
