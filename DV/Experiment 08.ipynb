# Time Series Forecasting and Trend Lines



**Aim:** Analyze monthly sales, calculate a moving average, fit a linear trend line, and forecast six months ahead. Simulated data is used for teaching, not production forecasting.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
np.random.seed(7)
dates = pd.date_range("2022-01-01", periods=48, freq="MS")
t = np.arange(len(dates))
values = 1000 + 18*t + 120*np.sin(2*np.pi*t/12) + np.random.normal(0,55,len(t))
ts = pd.DataFrame({"Date":dates,"Sales":values})
display(ts.head())
```

```python
ts["3_Month_MA"] = ts.Sales.rolling(3).mean()
model = LinearRegression().fit(t.reshape(-1,1), ts.Sales)
ts["Trend"] = model.predict(t.reshape(-1,1))
plt.figure(figsize=(11,5))
plt.plot(ts.Date,ts.Sales,marker="o",label="Actual sales")
plt.plot(ts.Date,ts["3_Month_MA"],label="3-month moving average")
plt.plot(ts.Date,ts.Trend,"--",label="Linear trend")
plt.title("Monthly Sales, Moving Average and Trend"); plt.xlabel("Month"); plt.ylabel("Sales")
plt.legend(); plt.tight_layout(); plt.show()
```

```python
future_dates = pd.date_range(ts.Date.iloc[-1] + pd.offsets.MonthBegin(1), periods=6, freq="MS")
future_t = np.arange(len(ts),len(ts)+6)
forecast = model.predict(future_t.reshape(-1,1))
forecast_df = pd.DataFrame({"Date":future_dates,"Forecast Sales":forecast})
display(forecast_df.round(2))
plt.figure(figsize=(11,5)); plt.plot(ts.Date,ts.Sales,label="Historical sales")
plt.plot(future_dates,forecast,marker="o",linestyle="--",label="6-month forecast")
plt.title("Sales Forecast — Linear Regression"); plt.xlabel("Month"); plt.ylabel("Sales")
plt.legend(); plt.tight_layout(); plt.show()
```

**Observations:** A moving average smooths short-term fluctuations. A linear trend summarizes the overall direction and the forecast extends that straight-line trend. It does not explicitly model seasonality, holidays, or prediction uncertainty.
