# Pie and Donut Charts for Category Proportion



**Aim:** Display proportions with pie and donut charts and analyze medal share by country. Medal counts are illustrative sample values, not official Olympic statistics.

```python
import pandas as pd
import matplotlib.pyplot as plt
medals = pd.DataFrame({
 "Country":["Country A","Country B","Country C","Country D","Country E"],
 "Gold":[38,29,21,17,13], "Silver":[32,25,24,18,15], "Bronze":[25,20,19,22,16]
})
medals["Total Medals"] = medals[["Gold","Silver","Bronze"]].sum(axis=1)
display(medals)
```

```python
plt.figure(figsize=(7,7))
plt.pie(medals["Total Medals"], labels=medals.Country, autopct="%1.1f%%", startangle=90)
plt.title("Share of Total Medals by Country"); plt.axis("equal"); plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.pie(medals["Total Medals"], labels=medals.Country, autopct="%1.1f%%", startangle=90,
        wedgeprops={"width":0.42,"edgecolor":"white"})
plt.title("Olympic Medal Share — Donut Chart"); plt.axis("equal"); plt.show()
```

```python
medals.set_index("Country")[["Gold","Silver","Bronze"]].plot(
    kind="bar", stacked=True, figsize=(9,5), title="Medal Composition by Country")
plt.ylabel("Number of medals"); plt.xticks(rotation=0); plt.tight_layout(); plt.show()
```

**Observations:** Pie and donut charts show each country's share of the total. The stacked bar chart compares medal composition. Pie charts are most useful with a small number of categories that form a meaningful whole.\n\n**Result:** Pie, donut, and stacked bar charts were created.
