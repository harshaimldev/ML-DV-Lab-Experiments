# Getting Started with Power BI / Tableau



**Aim:** Import and explore a dataset, inspect tabular data, and create basic bar and line charts. In Power BI use Home → Get Data → Text/CSV; in Tableau use Connect → Text File.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.frame
df["species"] = df["target"].map(dict(enumerate(iris.target_names)))
df = df.drop(columns="target")
print("Shape:", df.shape)
display(df.head())
display(df.describe())
display(df.dtypes)
```

```python
display(df.groupby("species").mean(numeric_only=True).round(2))
df["species"].value_counts().plot(kind="bar", title="Iris Samples by Species", rot=0)
plt.ylabel("Count"); plt.tight_layout(); plt.show()
```

```python
df.loc[:29, "sepal length (cm)"].plot(kind="line", marker="o", title="Sepal Length — First 30 Samples", figsize=(9,4))
plt.xlabel("Sample index"); plt.ylabel("Sepal length (cm)"); plt.tight_layout(); plt.show()
```

**Observations:** The table shows individual records and data types. The bar chart compares species counts, while the line chart displays a measurement across sample order.\n\n**Result:** The Iris dataset was imported, explored, and visualized with table, bar, and line views.
