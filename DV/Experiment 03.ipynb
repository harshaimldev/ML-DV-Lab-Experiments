# Creating Bar and Line Charts for Series Data



**Aim:** Compare category-wise sales with a bar chart and show sales trends over time with a line chart. Labels and hover tooltips are included.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
np.random.seed(42)
n = 500
dates = pd.date_range("2024-01-01", periods=180, freq="D")
sales = pd.DataFrame({
    "Order Date": np.random.choice(dates, n),
    "Region": np.random.choice(["North","South","East","West"], n),
    "Category": np.random.choice(["Furniture","Office Supplies","Technology"], n),
    "Sub-Category": np.random.choice(["Chairs","Phones","Binders","Tables","Accessories"], n),
    "Sales": np.round(np.random.gamma(4,120,n),2),
    "Profit": np.round(np.random.normal(35,55,n),2),
    "Quantity": np.random.randint(1,10,n)
}).sort_values("Order Date").reset_index(drop=True)
display(sales.head())
```

```python
category_sales = sales.groupby("Category")["Sales"].sum().sort_values(ascending=False)
ax = category_sales.plot(kind="bar", figsize=(8,4), title="Total Sales by Category", rot=0)
ax.set_xlabel("Category"); ax.set_ylabel("Sales")
for bars in ax.containers: ax.bar_label(bars, fmt="%.0f")
plt.tight_layout(); plt.show()
```

```python
daily = sales.groupby("Order Date")["Sales"].sum().sort_index()
plt.figure(figsize=(11,4)); plt.plot(daily.index, daily.values, marker=".")
plt.title("Daily Sales Trend"); plt.xlabel("Order Date"); plt.ylabel("Sales")
plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

```python
import plotly.express as px
trend = sales.groupby(["Order Date","Category"], as_index=False)["Sales"].sum()
px.line(trend, x="Order Date", y="Sales", color="Category", title="Daily Sales by Category",
        hover_data={"Sales":":,.2f"}).show()
```

**Observations:** The bar chart ranks categories by sales. The line chart shows day-to-day variation. Hover tooltips reveal exact values without crowding the plot.\n\n**Result:** Bar, line, and interactive charts were created.
