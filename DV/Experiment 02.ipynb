# Data Cleaning and Transformation



**Aim:** Handle missing values, rename columns, correct data types, and create calculated fields. This Pandas workflow demonstrates transformations similar to Power Query or Tableau Data Interpreter.

```python
import pandas as pd
import numpy as np
raw = pd.DataFrame({
 "Order Date":["2024-01-01","2024-01-02",None,"2024-01-04","2024-01-05"],
 "Product Name":["Laptop","Chair","Phone",None,"Desk"],
 "Sales Amount":[50000,4500,18000,np.nan,9000],
 "Quantity":[1,2,1,3,np.nan],
 "Region ":["North","South","East","West",None]
})
display(raw)
display(raw.isna().sum())
```

```python
df = raw.rename(columns={"Order Date":"order_date","Product Name":"product_name","Sales Amount":"sales","Region ":"region"})
df.columns = df.columns.str.strip().str.lower().str.replace(" ","_")
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")
df["product_name"] = df["product_name"].fillna("Unknown")
df["region"] = df["region"].fillna("Not specified")
df["sales"] = df["sales"].fillna(df["sales"].median())
df["quantity"] = df["quantity"].fillna(df["quantity"].median()).astype(int)
df["total_value"] = df["sales"] * df["quantity"]
df["average_unit_price"] = df["sales"] / df["quantity"]
display(df); display(df.dtypes); display(df.isna().sum())
```

**Observations:** Numeric missing values were replaced with the median and missing text values were labelled. Dates were converted to datetime. Median imputation is only one possible strategy; select a method based on the data.\n\n**Result:** The dataset was cleaned and calculated columns were added.
