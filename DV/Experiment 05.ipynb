# Geo-Visualization with Maps



**Aim:** Visualize country-wise sales using a filled choropleth map and a symbol map. The values below are illustrative sample data, not real business results.

```python
import pandas as pd
import plotly.express as px
geo = pd.DataFrame({
 "Country":["India","United States","Canada","United Kingdom","Germany","Australia","Japan","Brazil"],
 "Sales":[82000,125000,54000,68000,72000,61000,77000,43000]
})
display(geo)
```

```python
px.choropleth(geo, locations="Country", locationmode="country names", color="Sales",
                hover_name="Country", color_continuous_scale="Blues",
                title="Country-wise Sales — Choropleth").show()
```

```python
coords = pd.DataFrame({
 "Country":["India","United States","Canada","United Kingdom","Germany","Australia","Japan","Brazil"],
 "lat":[20.59,37.09,56.13,55.38,51.17,-25.27,36.20,-14.24],
 "lon":[78.96,-95.71,-106.35,-3.44,10.45,133.78,138.25,-51.93],
 "Sales":[82000,125000,54000,68000,72000,61000,77000,43000]
})
px.scatter_geo(coords, lat="lat", lon="lon", size="Sales", color="Sales",
               hover_name="Country", projection="natural earth",
               title="Country-wise Sales — Symbol Map").show()
```

**Observations:** Choropleth colour intensity compares values across geographic areas. Symbol-map marker size represents sales. Geographic fields should be categorized correctly and ambiguous locations should include state/country context.\n\n**Result:** Filled and symbol maps were created.
