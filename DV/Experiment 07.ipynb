# Heatmaps and Tree Maps



**Aim:** Create a correlation heatmap for numerical variables and a hierarchical treemap for category/subcategory sales.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
np.random.seed(42)
n = 500
dates = pd.date_range("2024-01-01", periods=180, freq="D")
sales = pd.DataFrame({
    "Order Date": np.random.choice(dates, n),
    "Region": np.random.choice(["North","South","East","West"], n),
    "Category": np.random.choice(["Furniture","Office Supplies","Technology"], n),
    "Sub-Category": np.random.choice(["Chairs","Phones","Binders","Tables","Accessories"], n),
    "Sales": np.round(np.random.gamma(4,120,n),2),
    "Profit": np.round(np.random.normal(35,55,n),2),
    "Quantity": np.random.randint(1,10,n)
}).sort_values("Order Date").reset_index(drop=True)
display(sales.head())
```

```python
corr = sales[["Sales","Profit","Quantity"]].corr(method="pearson")
display(corr.round(2))
plt.figure(figsize=(6,4)); sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1, fmt=".2f")
plt.title("Correlation Heatmap"); plt.tight_layout(); plt.show()
```

```python
import plotly.express as px
tree = sales.groupby(["Category","Sub-Category"], as_index=False)["Sales"].sum()
px.treemap(tree, path=["Category","Sub-Category"], values="Sales", color="Sales",
           color_continuous_scale="Blues", title="Sales by Category and Sub-Category").show()
```

**Observations:** Correlation ranges from -1 to +1 and measures linear association; it does not prove causation. Treemap rectangle area represents sales contribution and nested rectangles show hierarchy.\n\n**Result:** A correlation heatmap and hierarchical treemap were created.
